```python

# EXPLORATORY DATA ANALYSIS PROJECT

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder

# 2. Load Dataset
# Replace with your CSV file name
file_path = "data.csv"
if not os.path.exists(file_path):
    # Fallback to a built-in Colab sample dataset if "data.csv" is not found
    file_path = "sample_data/california_housing_train.csv"
    print(f"'data.csv' not found. Falling back to sample dataset: {file_path}\n")

df = pd.read_csv(file_path)

print("FIRST FIVE ROWS")
print(df.head())

# 3. Dataset Information
print("\nDATASET INFORMATION")
print(df.info())

print("\nSTATISTICAL SUMMARY")
print(df.describe())

# 4. Check Missing Values
print("\nMISSING VALUES")
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Fill missing numerical values
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# 5. Statistical Analysis
print("\nMEAN")
print(df.mean(numeric_only=True))

print("\nMEDIAN")
print(df.median(numeric_only=True))

print("\nSTANDARD DEVIATION")
print(df.std(numeric_only=True))

# 6. Data Visualization

# Histogram
df.hist(figsize=(12, 8), bins=20)
plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()

# Correlation Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()

# Boxplot
plt.figure(figsize=(10, 5))
sns.boxplot(data=df.select_dtypes(include=np.number))
plt.title("Outlier Detection Using Boxplot")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 7. Identify Correlations
corr = df.corr(numeric_only=True)

print("\nCORRELATION MATRIX")
print(corr)

# Find highly correlated variable pairs
high_corr = corr.abs().unstack()
high_corr = high_corr[
    (high_corr > 0.7) & (high_corr < 1.0)
]

print("\nHIGHLY CORRELATED VARIABLES")
print(high_corr)

# 8. Identify Key Influencing Factors
# Select the last numerical column as target
numeric_df = df.select_dtypes(include=np.number)

if numeric_df.shape[1] >= 2:
    target = numeric_df.columns[-1]

    X = numeric_df.drop(columns=[target])
    y = numeric_df[target]

    model = RandomForestRegressor(
        n_estimators=100,
        random_state=42
    )

    model.fit(X, y)

    importance = pd.Series(
        model.feature_importances_,
        index=X.columns
    ).sort_values(ascending=False)

    print("\nKEY INFLUENCING FACTORS")
    print(importance)

    # Feature Importance Visualization
    importance.plot(
        kind="bar",
        figsize=(8, 5),
        color="skyblue"
    )
    plt.title("Feature Importance")
    plt.ylabel("Importance Score")
    plt.tight_layout()
    plt.show()

# 9. Generate Structured Analytical Report
with open("EDA_Report.txt", "w") as file:

    file.write("EXPLORATORY DATA ANALYSIS REPORT\n")
    file.write("=" * 40 + "\n\n")

    file.write("Dataset Shape:\n")
    file.write(str(df.shape) + "\n\n")

    file.write("Statistical Summary:\n")
    file.write(str(df.describe()) + "\n\n")

    file.write("Missing Values:\n")
    file.write(str(df.isnull().sum()) + "\n\n")

    file.write("Correlation Analysis:\n")
    file.write(str(corr) + "\n\n")

    if numeric_df.shape[1] >= 2:
        file.write("Important Influencing Factors:\n")
        file.write(str(importance) + "\n")

print("\nANALYSIS COMPLETED SUCCESSFULLY!")
print("EDA Report saved as EDA_Report.txt")
```
